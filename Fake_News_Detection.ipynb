```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.metrics import accuracy_score

print("1. Loading dataset...")
url = "https://raw.githubusercontent.com/lutzhamel/fake-news/master/data/fake_or_real_news.csv"
df = pd.read_csv(url)
print(f"-> Dataset loaded successfully with shape: {df.shape}")

print("\n2. Preparing and splitting data...")
labels = df['label'].values
x_train, x_test, y_train, y_test = train_test_split(
    df['text'], labels, test_size=0.2, random_state=7
)

print("\n3. Vectorizing text using TF-IDF...")
tfidf_vectorizer = TfidfVectorizer(stop_words='english', max_df=0.7)
tfidf_train = tfidf_vectorizer.fit_transform(x_train)
tfidf_test = tfidf_vectorizer.transform(x_test)

print("\n4. Training the PassiveAggressiveClassifier model...")
pac = PassiveAggressiveClassifier(max_iter=50)
pac.fit(tfidf_train, y_train)

# Evaluate and print accuracy
y_pred = pac.predict(tfidf_test)
score = accuracy_score(y_test, y_pred)
print(f"-> Model trained successfully! Accuracy: {round(score * 100, 2)}%")

print("\n--------------------------------------------------")
print("5. Interactive News Checker Ready!")
print("--------------------------------------------------")

# Interactive User Input Function
def check_user_news():
    user_input = input("Enter the news text you want to check: \n> ")

    if not user_input.strip():
        print("Please enter some valid text.")
        return

    # Transform input and predict
    transformed_input = tfidf_vectorizer.transform([user_input])
    prediction = pac.predict(transformed_input)

    print("\n--- Result ---")
    print(f"The given news is predicted to be: **{prediction[0]}**")

# Run the input prompt
check_user_news()
```
